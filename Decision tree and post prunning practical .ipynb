```python
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
from sklearn.datasets import load_iris
iris =load_iris()
iris.data
```

```python
iris.target
```

```python
import seaborn as sns
df=sns.load_dataset("iris")
df
```

```python
x=df.iloc[:,:-1]
y=iris.target
```

```python
#train test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
#in decision tree classifir we usr parameter max_depth
dtc=DecisionTreeClassifier(max_depth=4,splitter='best',criterion='gini',max_features='sqrt')
```

```python
dtc.fit(x_train,y_train)
```

```python
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(dtc,filled=True)
plt.show()
```

```python
y_pred=dtc.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score
accurasy=accuracy_score(y_pred,y_test)
accurasy
```

this is a part of pre prunning in this we use parameter , we can say we find the parameter that are good fit for the model in this before apply  and chose best params after this we perfom decision treee classifiers 

```python
parameter={
"criterion":['gini', 'entropy', 'log_loss'], 
"splitter":["best", "random"], 
"max_depth":[1,2,3,4,5],
"max_features":["auto", "sqrt", "log2"]

}
```

```python
from sklearn.model_selection import GridSearchCV
trees=DecisionTreeClassifier()
cv=GridSearchCV(trees,param_grid=parameter,cv=5,scoring='accuracy')
```

```python
cv.fit(x_train,y_train)
```

```python
cv.best_params_
```
